# Data preprocessing

## 1. Handle Missing Values

```python
df["Income"]=df["Income"].fillna(df["Income"].median())
```

```python
df.head()
```

```python
df.isnull().sum()
```

## 2. Feature Engineering

```python
# Age
df["Age"]= 2026-df["Year_Birth"]
```

```python
df.head()
```

```python
# customer Joining date

df["Dt_Customer"]=pd.to_datetime(df["Dt_Customer"],dayfirst=True)  # pandas expect it in mm-dd-yyyy format

reference_date=df["Dt_Customer"].max()

df["Customer_Tenure_Days"]=(reference_date-df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
# Spending

df["Total_spending"]=df["MntWines"]+df["MntFruits"]+df["MntMeatProducts"]+df["MntFishProducts"]+df["MntSweetProducts"]+df["MntGoldProds"]
```

```python
df.head()
```

```python
df["Total_Children"]=df["Kidhome"]+df["Teenhome"]
```

```python
df.head()
```

```python
# Education

df["Education"].value_counts()

# Undergrad,Grad,Postgrad
df["Education"]=df["Education"].replace({
    "Basic":"Undergraduate","2n Cycle": "Undergraduate",
    "Graduation":"Graduate",
    "Master":"Postgraduate","PhD":"Postgraduate"
})
```

```python
df["Education"].value_counts()
```

```python
# Marital Status

df["Marital_Status"].value_counts()
```

```python
df["Living_With"]=df["Marital_Status"].replace({
    "Married":"Partner","Together": "Partner",
    "Single":"Alone","Divorced":"Alone",
    "Widow":"Alone","Absurd":"Alone","YOLO":"Alone"
})
```

```python
df["Living_With"].value_counts()
```

## Drop Columns

```python
cols=["ID","Year_Birth","Marital_Status","Kidhome","Teenhome","Dt_Customer"]
spending_cols=["MntWines","MntFruits","MntMeatProducts","MntFishProducts","MntSweetProducts","MntGoldProds"]


cols_to_drop=cols+spending_cols

df_cleaned=df.drop(columns=cols_to_drop)
```

```python
df_cleaned.shape
```

```python
# from these graphs we can potentially identify the possible outliers and treat them accordingly
```

# Encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder()

cat_cols=["Education","Living_With"]
enc_cols=ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df=pd.DataFrame(enc_cols.toarray(),columns=ohe.get_feature_names_out(cat_cols),index=df_cleaned.index)
# This code converts the One-Hot Encoded data into a pandas DataFrame (enc_df) with proper column names and the same row indexes as df_cleaned.
# here drop_first=True not used check why
```

```python
enc_df.head()
```

```python
df_encoded=pd.concat([df_cleaned.drop(columns=cat_cols),enc_df],axis=1)
```

```python
# 3D

pca=PCA(n_components=3)

X_pca=pca.fit_transform(X_scaled)
pca.explained_variance_ratio_
```

```python
# this captures 23+11+10=44% of information 
```

```python
# plot
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")  # general method for 3D graph 
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2])

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("3D projection")
```

# Analyze K value

### 1. Elbow Method

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss=[]
for k in range(1,11):
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee=KneeLocator(range(1,11),wcss,curve="convex",direction="decreasing")
optimal_k=knee.elbow
```

```python
print("best k = ",optimal_k)
```

```python
# plot

plt.plot(range(1,11),wcss,marker='o')
plt.xlabel("K")
plt.ylabel("WCSS") 
```

## Silhouette Score

```python
from sklearn.metrics import silhouette_score

scores=[]

for k in range(2,11):
    kmeans=KMeans(n_clusters=k,random_state=42) 
    labels=kmeans.fit_predict(X_pca)    # Stores the cluster number assigned to each data point by K-Means
    score=silhouette_score(X_pca,labels)  #  # Stores the Silhouette Score, which measures how well the clusters are separated; a score closer to +1 generally indicates better-defined clusters.
    scores.append(score) 

# plot
plt.plot(range(2,11),scores,marker='o')
plt.xlabel("K")
plt.ylabel("Silhouette score") 
```

```python
# combined plot

k_range=range(2,11)

fig,ax1=plt.subplots(figsize=(8,6))

ax1.plot(k_range,wcss[:len(k_range)],marker="o",color="blue")
ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2=ax1.twinx()
ax2.plot(k_range,scores[:len(k_range)],marker="x",color="red",linestyle="--")
ax2.set_ylabel("SS")
```

# Clustering

```python
# K_means 

kmeans=KMeans(n_clusters=4,random_state=42)
labels_kmeans=kmeans.fit_predict(X_pca)
```

```python
fig=plt.figure(figsize=(8,6))

ax=fig.add_subplot(111,projection="3d")

ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels_kmeans)
```

```python
# Agglomerative Clustering
from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf=AgglomerativeClustering(n_clusters=4,linkage="ward")
labels_agg=agg_clf.fit_predict(X_pca)
```

```python
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels_kmeans)
```

```python
# Aggomerative Clustering
from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf=AgglomerativeClustering(n_clusters=4,linkage="ward") # Ward linkage merges, at each step, the two clusters whose combination increases the total spread (variance) the least, so you end up with tight, compact groups.
labels_agg=agg_clf.fit_predict(X_pca) 
```

```python
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels_agg)
```
